# Household + weather data

Result of `python Combination/merge_weather.py`: one row per household and 15-minute timestamp.

```python
from pathlib import Path

import pandas as pd

PATH = Path.cwd().parent / "data" / "preprocessed" / "household_weather.parquet"
if not PATH.exists():  # notebook started from the repository root
    PATH = Path.cwd() / "data" / "preprocessed" / "household_weather.parquet"

df = pd.read_parquet(PATH)
print(f"{len(df):,} rows, {df['Household_ID'].nunique()} households, {df['Weather_ID'].nunique()} stations")
```
